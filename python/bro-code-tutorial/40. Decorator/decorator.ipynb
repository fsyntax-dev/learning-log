
## Decorator



```python
#Assign function to variable

def plus_one(number):
    return number + 1

add_one = plus_one

add_one(6)
```

```python
#Function Inside Function

def plus_one(number):
    def add_one(number):
        return number + 1
    
    result = add_one(number)
    return result

plus_one(7)
```

```python
#Passing Function as Arguements to Other Functions
def plus_one(number):
    return number + 1

def function_call(function):
    number_to_add = 5
    return function(number_to_add)

#Function passed as (paramenter)
function_call(plus_one)
```

```python
def outer_function():
    def inner_function1():
        return "Hello from inner!"
    
    def inner_function2():
        return "Hello from inside"
    
    return inner_function1(), inner_function2()

myfunction = outer_function()

print(myfunction)
```

```python
def upper_decorator(function):
    def wrapper():
        func = function()
        make_uppercase = func.upper()
        return make_uppercase
    
    return wrapper


def say_hi():
    return "Hello There"

decorate = upper_decorator(say_hi)

decorate()
```

```python
@upper_decorator
def say_hi():
    return "Hello There"

say_hi()
```
