## Stochastic Process

Stochastic Process - a study on randomness of a variable

## Random Walk

```python
n_steps = 252
np.random.seed(40)

steps = np.random.choice([-1, 1], size=n_steps) 
walk = np.cumsum(steps) # sum of all step
```

```python
plt.figure(figsize=(12, 5))
plt.plot(walk, color='steelblue', linewidth=1.5)
plt.axhline(0, color='red', linestyle='--', alpha=0.5)
plt.title("Random Walk (252 steps)")
plt.xlabel("Steps")
plt.ylabel("Position")
plt.grid(True, alpha=0.3)
plt.show()
```
