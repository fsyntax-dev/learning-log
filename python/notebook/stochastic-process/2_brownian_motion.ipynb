```python
import numpy as np
import matplotlib.pyplot as plt
```

Random Walk - fixed step (+1, -1, +1, +1)
BM          - continuous step (+0.5, +3, -0.025, +3.25)
            - BM dont have drift

```python
n_steps = 252 
dt = 1/252 

dW = np.random.normal(0, np.sqrt(dt), n_steps)
W = np.cumsum(dW)   
```

```python
plt.figure(figsize=(12, 5))
plt.plot(W, color='steelblue', linewidth=1.5)
plt.axhline(0, color='red', linestyle='--', alpha=0.5)
plt.title("Brownian Motion (252 steps)")
plt.xlabel("Steps")
plt.ylabel("W(t)")
plt.grid(True, alpha=0.3)
plt.show()
```
