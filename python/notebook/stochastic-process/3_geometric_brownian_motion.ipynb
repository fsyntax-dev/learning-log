```python
import numpy as np
import matplotlib.pyplot as plt
```

GBM     - randomness + drift
        - drift is just a direction
            +drift = upward
            -drift = downward
            0drift = no direction 


epsilon - random number from normal distribution

dt = T / steps
randomness = sigma x epsilon x sqrt(dt)
drif = (mean - sigma^2/2) x dt

St+1 = St x e^(drift + random)


```python
S0 = 300
mu = 0.20
sigma = 0.40
T = 1
n_steps = 252

dt = T / n_steps
```

```python
prices = [S0]

for step in range(n_steps):
    drift = (mu - sigma**2 /2) * dt
    random = sigma * np.random.normal(0, 1) * np.sqrt(dt)
    next_price = prices[-1] * np.exp(drift + random)
    prices.append(next_price)
```

```python
plt.figure(figsize=(12, 6))
plt.plot(prices, linewidth=2)
plt.title("Simulation of Asset Price with Geometric Brownian Motion")
plt.xlabel("Days")
plt.ylabel("Price ($)")
plt.grid(True, alpha=0.3)
plt.show()
```
