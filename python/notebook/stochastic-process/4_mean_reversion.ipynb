```python
import numpy as np
import matplotlib.pyplot as plt
```

## Mean Reversion (Ornstein-Uhlenbeck)

mean reversion - randomness + gravitates back to mean

Ornstein-Uhlenbeck - particle movement in fluid - partice pushed around but always return back to equilibirum

dS = k(theta - S) x dt + sigma x epsilon x sqrt(dt)

dS      = delta in price
kappa   = speed of reversion
theta   = mean
S       = current price
dt      = T/steps
sigma   = volatility
epsilon = random number from normal distribution


if sigma > kappa = randomness win
if sigma < kappa = reversion win


```python
S0 = 0.05
kappa = 3.0
theta = 0.05
sigma = 0.01
T = 10
n_steps = 252 * T

dt = T/n_steps
```

```python
prices = [S0]

for step in range(n_steps):
    dS = kappa * (theta - prices[-1]) * dt + sigma * np.random.normal(0, 1) * np.sqrt(dt)
    next_price = prices[-1] + dS
    prices.append(next_price)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(prices)

ax = plt.gca()
ax.axhline(theta, color="red", linestyle="--")
ax.set_xlabel("Date", loc="center")
ax.set_ylabel("Rate", loc="center")
ax.set_title("Mean Reversion — Interest Rate Simulation")

plt.show()
```
