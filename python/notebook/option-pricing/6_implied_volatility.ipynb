```python
import numpy as np
from scipy.stats import norm 
from scipy.optimize import brentq
```

## Implied Volatility

IV -  volatility can be translated by the premium

in real market - there is live feed of premium price 
    we just reverse engineer the IV/guessing around with our model to come out same price as live price
    so our model it will tells the sigma

the guessing thingy is called Root Finding
Brent's Method is one of root-finding algorithm

```python
S = 100
K = 100
T = 1
r = 0.05
```

```python
# we use Black Scholes to match the market price
def bsm_call(S, K, T, r, sigma):
    d1 = (np.log(S/K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    return S * norm.cdf(d1) - K * np.exp(-r * T) * norm.cdf(d2)
```

```python
market_price = 10.45

iv = brentq(lambda sigma: bsm_call(S, K, T, r, sigma) - market_price, 0.001, 10)

print(f"Market Price        : ${market_price}")
print(f"Implied Volatility  : {iv:.4f} ({iv*100:.2f}%)")
```
