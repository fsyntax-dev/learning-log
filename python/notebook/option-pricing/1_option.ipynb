# Option

option - a contract that gives right to buy asset at predetermined price

call - right to buy
put  - right to sell

underlying (s)      - current price 
strike price (k)    - price stated in contract
premium             - price of contract
expiry (t)          - date contract expires

european option     - exercise only at expiry date
american option     - exercise anytime before expiry date


## Moneyness

moneyness - relantionship of S and K

In The Money    - profit
At The Money    - breakeven
Out The Money   - loss

payoff - amount receive when option exercised
profit - amount you make after deduct premium

```python
S = 500
K = 300
premium = 10

call_payoff = max(S - K, 0)
call_profit = call_payoff - premium

print(f"Profit: {call_profit}")

```
