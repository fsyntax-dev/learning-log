```python
import numpy as np
from scipy.stats import norm 
```

## Greeks

greek -  measure option price (premium) sensitivity based on

delta   - change in asset price
gamma   - change in delta
theta   - change in time
vega    - change in volatility
rho     - change in risk free rate

```python
S = 120
K = 100
T = 1
r = 0.05
sigma = 0.2
```

CDF - Cumulative Distribution Function
PDF - Probability Density Function

### Delta

delta - if asset price move by $1, how much does option price(premium) change

Δcall = N(d1)
Δput = N(d1)- 1

```python
d1 = (np.log(S/K) + (r + 0.5 * sigma **2) * T) / (sigma * np.sqrt(T))
d2 = d1 - sigma * np.sqrt(T)

Nd1 = norm.cdf(d1)
print(f"Delta : {Nd1:.4f}")

print(f"When S move by $1, Premium will add {Nd1:.4f}")
```

### Gamma

gamma - changes of delta
      - current delta - previous delta (but we doesnt use subtraction)

gammma = N'(d1) / S * sigma * sqrt(T)      

```python
gamma = norm.pdf(d1) / (S * sigma * np.sqrt(T))
print(f"Gamma : {gamma:.4f}")
print(f"When S moves $1, Delta will change by {gamma:.4f}")
```

### Theta

theta - if 1 day pass - how much the premium change

theta = -S * norm.pdf(d1) * sigma / 2 * sqrt(T)  - risk free rate * K * np.exp(-r x T) * norm.cdf(d2)

theta figure is for full year, divide by 365 to get daily figure

time decay - period shrinking - less opportunity - premium drop everyday - theta is negative for buyer
time decay favor seller

```python
theta = (-S * norm.pdf(d1) * sigma / (2 * np.sqrt(T))) - (r * K * np.exp(-r * T) * norm.cdf(d2))

print(f"Theta : {theta:.4f}")
print(f"Every 1 day passes, premium changes by ${theta/365:.4f}")
```

### Vega

vega - if volatility (sigma) by 1%, how much does the premium change

high volatility - higher change of moves - premium worth more

```python
vega = S * norm.pdf(d1) * np.sqrt(T)

print(f"Vega : {vega:.4f}")
print(f"If volatility changes by by 1%, premium changes b ${vega/100:.4f}")
```

### Rho

rho - if risk free rate changes by 1%, how much does premium change

r goes up -> call worth more, put worth less

rho_call = K * T * np.exp(-r x T) X norm.cdf(d2)
rho_put  = -K * T * np.exp(-r x T) X norm.cdf(-d2)

```python
rho_call = K * T * np.exp(-r * T) * norm.cdf(d2)
rho_put = -K * T * np.exp(-r * T) * norm.cdf(-d2)

print(f"Rho Call : {rho_call:.4f}")
print(f"Rho Put  : {rho_put:.4f}")
print(f"If r changes by 1%, call premium changes by ${rho_call/100:.4f}")
print(f"If r changes by 1%, put premium changes by ${rho_put/100:.4f}")
```
