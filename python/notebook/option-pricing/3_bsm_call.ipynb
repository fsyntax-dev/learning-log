```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot
from scipy.stats import norm
from datetime import datetime
```

```python
tsla = yf.download("TSLA", period="5y")["Close"]["TSLA"]
```

```python
S = tsla.iloc[-1]

print(f"S: {S:.2f}")
```

```python
tsla = yf.Ticker("TSLA")
date = tsla.options[10]
expiry = datetime.strptime(date, "%Y-%m-%d")
today = datetime.today()

T = (expiry - today).days / 365
print(f"T: {T:.4f}")
```

```python
chain = tsla.option_chain(date)

call = chain.calls

K = call["strike"]
sigma = call["impliedVolatility"]
```

```python
K = K.iloc[8]
sigma = sigma.iloc[8]

print(f"K       : {K}")
print(f"sigma   : {sigma:.4f}")
```

```python
r = yf.download("^TNX", period="5d")["Close"]["^TNX"]
r = r.iloc[-1]
r = r / 100

print(f"r : {r:.4f}")
```

```python
d1 = (np.log(S/K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
d2 = d1 - sigma * np.sqrt(T)

Nd1 = norm.cdf(d1)
Nd2 = norm.cdf(d2)

print(f"N(d1) : {Nd1:.4f}")
print(f"N(d2) : {Nd2:.4f}")
```

```python
call_price = S * Nd1 - K * np.exp(-r * T) * Nd2
print(f"Call Price: $ {call_price:.2f}")
```
