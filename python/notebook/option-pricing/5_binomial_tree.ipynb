```python
import numpy as np
```

## Binomial Tree

Binomial Tree Model - alternative ways to price premium
                    - build a tree of possible stock price until expiry

Build tree of possible S values -> Calc Payoff -> Calc Premium

     Su(premium goes up)
    /
S
    \
     Sd(premium goes down)

u = up factor
d = down factor     

```python
S = 100
K = 100
T = 1
r = 0.05
sigma = 0.2
```

### Factor

```python
u = np.exp(sigma * np.sqrt(T))
d = np.exp(-sigma * np.sqrt(T))

print(f"Up factor   : {u:.4f}")
print(f"Down factor : {d:.4f}")
```

```python
Su = S * u
Sd = S * d

print(f"Su : ${Su:.4f}")
print(f"Sd : ${Sd:.4f}")
```

### Payoff

```python
payoff_u = max(Su - K, 0)
payoff_d = max(Sd - K, 0)

print(f"Payoff up   : {payoff_u:.4f}")
print(f"Payoff down : {payoff_d:.4f}")
```

### Risk Neutral Probability

RNP - to find probability of factor
p = np.exp(r * T) - down factor / up factor - down factor

```python
p = (np.exp(r * T) - d) / (u - d)
print(f"Risk neutral probability : {p:.4f}")
```

### Premium

premium = np.exp(-r x T) * (RNP * payoff_up + (1 + RNP) * payoff_down)

```python
premium = np.exp(-r * T) * (p * payoff_u + (1 - p) * payoff_d)
print(f"Premium : ${premium:.4f}")
```

### Multi Step Binomial Tree

```python
n = 1000  # number of steps
dt = T / n  # time per step

u = np.exp(sigma * np.sqrt(dt))
d = np.exp(-sigma * np.sqrt(dt))
p = (np.exp(r * dt) - d) / (u - d)

# Build stock price tree at expiry
ST = np.array([S * (u**j) * (d**(n-j)) for j in range(n+1)])

# Calculate payoff at expiry
payoffs = np.maximum(ST - K, 0)

# Work backwards
for i in range(n):
    payoffs = np.exp(-r * dt) * (p * payoffs[1:] + (1-p) * payoffs[:-1])

print(f"Binomial Tree Premium (1000 steps) : ${payoffs[0]:.4f}")
```
