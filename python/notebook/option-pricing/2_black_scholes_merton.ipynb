```python
import numpy as np
from scipy.stats import norm
```

## Black Scholes Merton

Black Scholes Model - a math model to calc the fair price of premium right now

S - current asset price
K - price of the contract
T - time until expiry date
r - risk free rate
σ - volatility of the asset

## Call

C = S x N(d1​) − K x e−rT x N(d2​)

```python
S = 100
K = 100
T = 1
r = 0.05
sigma = 0.2
```

```python
d1 = (np.log(S/K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
print(f"{d1:.4f}")

d2 = d1 - sigma * np.sqrt(T)
print(f"{d2:.4f}")
```

```python
Nd1 = norm.cdf(d1)
Nd2 = norm.cdf(d2)

print(f"N(d1): {Nd1:.4f}")
print(f"N(d2): {Nd2:.4f}")
```

```python
call_price = S * Nd1 - K * np.exp(-r * T) * Nd2
print(f"Call Price: $ {call_price:.4f}")
```

## Put

P = K x e-rt x N(-d2) - S x N(-d1)

```python
S = 100
K = 100
T = 1
r = 0.05
sigma = 0.2
```

```python
d1 = (np.log(S/K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
print(f"{d1:.4f}")

d2 = d1 - sigma * np.sqrt(T)
print(f"{d2:.4f}")
```

```python
Nd1 = norm.cdf(-d1)
Nd2 = norm.cdf(-d2)

print(f"N(d1): {Nd1:.4f}")
print(f"N(d2): {Nd2:.4f}")
```

```python
put_price = K * np.exp(-r * T) * Nd2 - S * Nd1
print(f"Put Price: $ {put_price:.4f}")
```
