```python
import pandas as pd
import numpy as np
import yfinance as yf
from scipy import stats
import matplotlib.pyplot as plt
```

## Volatility

Volatility - how much an asset move in a period

It just a standard deviation

```python
sp500 = yf.download("SPY", period="10y")["Close"]
sp500_returns = sp500["SPY"].pct_change(fill_method=None).dropna()

daily_vol = sp500_returns.std()
annual_vol = daily_vol* np.sqrt(252)

print(f"Daily Volatility    : {daily_vol*100:.2f} %")
print(f"Anual Volatility    : {annual_vol*100:.2f} %")
```

## Beta

Beta    - how much an asset move compared to index

It just a coefficient (slope)

if 1 % of Y = n% X moves

```python
tesla = yf.download("TSLA", period="10y")["Close"]
tesla_returns = tesla.pct_change(fill_method=None).dropna()

df_returns = pd.concat([sp500, tesla], axis=1).pct_change(fill_method=None).dropna()

covariance = df_returns["TSLA"].cov(df_returns["SPY"])
var = df_returns["SPY"].var() 

beta = covariance / var

print(f"Tesla Beta : {beta:.2f}")
```

```python
# linear regression to check coef with the beta

slope, intercept, r_value, p_value, std_err = stats.linregress(
    df_returns["SPY"],
    df_returns["TSLA"]
)

print(f"Slope       : {slope:.5f}")
print(f"Intercept   : {intercept:.5f}")
print(f"R-squared   : {r_value**2:.5f}")
print(f"P-value     : {p_value:.5f}")
```
