```python
import pandas as np
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt
```

above 1.0 is considered good

## Sharpe Ratio

Sharpe - return that you get for risk you took
Sharpe = Return - Risk Free Rate / Volatility


```python
sp500 = yf.download("SPY", period="10y")["Close"]["SPY"]
sp500_returns = sp500.pct_change(fill_method=None).dropna()

risk_free_rate = yf.download("^IRX", period="10y")["Close"].iloc[-1].iloc[0] / 100

cagr = (sp500.iloc[-1] / sp500.iloc[0]) ** (1/10) - 1
volatility = sp500_returns.std() * np.sqrt(252)

sharpe = (cagr - risk_free_rate) / volatility

```

```python
print(f"CAGR           : {float(cagr) * 100:.2f}%")
print(f"Risk Free Rate : {float(risk_free_rate)* 100:.2f}%")
print(f"Volatility     : {float(volatility) * 100:.2f}%")
print(f"Sharpe Ratio   : {float(sharpe):.2f}")
```

## Sortino Ratio

Sortino Ratio   - just a smarter version of Sharpe Ratio
                - it just excluded positive volatility

Sortino Ratio = Return - Risk Free Rate / Negative Volatility

```python
negative_returns = sp500_returns[sp500_returns < 0]
negative_volatility = negative_returns.std() * np.sqrt(252)

sortino = (cagr - risk_free_rate) / negative_volatility

print(f"Negative Volatility : {negative_volatility * 100:.2f}%")
print(f"Sortino Ratio       : {sortino:.2f}")
```

## Calmar Ratio

Calmar Ratio - return compared to max drawdown

Calmar Ratio = Return / Max DD

```python
rolling_peak = sp500.cummax()

drawdown = (sp500 - rolling_peak) / rolling_peak
max_drawdown = drawdown.min()

calmar = float(cagr) / abs(float(max_drawdown))

print(f"CAGR         : {float(cagr) * 100:.2f}%")
print(f"Max Drawdown : {float(max_drawdown) * 100:.2f}%")
print(f"Calmar Ratio : {calmar:.2f}")
```
