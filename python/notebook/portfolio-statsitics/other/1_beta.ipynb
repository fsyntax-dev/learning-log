```python
import pandas as pd
import yfinance as yf
from scipy import stats
import matplotlib.pyplot as plt
```

```python
assetOne = "AUDUSD=X"
assetTwo = "AUDJPY=X"
```

```python
assetOne = yf.download(assetOne, interval="15m", period="1mo", progress=False)["Close"]
assetTwo = yf.download(assetTwo, interval="15m", period="1mo", progress=False)["Close"]

assetOne_returns = assetOne.pct_change(fill_method=None)
assetTwo_returns = assetTwo.pct_change(fill_method=None)


df = pd.concat([assetOne_returns, assetTwo_returns], axis=1)
df.columns = ["Var X", "Var Y"]

df = df.dropna()

slope, intercept, r_value, p_value, std_err = stats.linregress(
    df["Var X"],
    df["Var Y"]
)

print(f"Slope       : {slope:.5f}")
print(f"Intercept   : {intercept:.5f}")
print(f"R-squared   : {r_value**2:.5f}")
print(f"P-value     : {p_value:.5f}")
```
