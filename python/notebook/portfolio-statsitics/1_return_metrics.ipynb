```python
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
```

## Total Return

```python
sp500 = yf.download("SPY", period="10y")["Close"]


start = sp500.iloc[0].iloc[0]
end = sp500.iloc[-1].iloc[0]  

total_return = ((end - start) / start) * 100

print(f"Start Price  : ${start:.2f}")
print(f"End Price    : ${end:.2f}")
print(f"Total Return : {total_return:.2f}%")
```

```python
sp500.plot(figsize=(10,5), title="SPY Daily Close Price")
plt.ylabel("Price")
plt.show()
```

## Daily Return

```python
daily_returns = sp500["SPY"].pct_change(fill_method=None).dropna()

print(f"Mean Daily Return : {daily_returns.mean():.4f}")
print(f"Std Daily Return  : {daily_returns.std():.4f}")
```

```python
daily_returns.plot(figsize=(10,5), title="SPY Daily Returns")
plt.axhline(0, color="red", linewidth=0.8, linestyle="--")
plt.ylabel("Daily Return")
plt.show()
```

## Compound Annual Growth Rate (CAGR)

CAGR - just like total return but with anuall compounding effect

```python
cagr = (end/start)**(1/10) - 1

print(f"Start Price : ${start:.2f}")
print(f"End Price   : ${end:.2f}")
print(f"CAGR        : {cagr * 100:.2f}%")
```
