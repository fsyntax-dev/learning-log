```python
import pandas as pd
import yfinance as yf
from scipy import stats
```

## Chi Square

Chi Square - answer for are two Categorical variable related?

```python
CRASH_METRICS = -0.03

sp500 = yf.download("^GSPC", period="10y")[["Close"]]
sp500.columns = ["SP500"]
sp500_returns = sp500["SP500"].pct_change(fill_method=None)

df = sp500_returns.to_frame()
df.columns = ["Returns"]

df["Day"] = df.index.dayofweek
df["Crash"] = (df["Returns"] < CRASH_METRICS).astype(int)

table = pd.crosstab(df["Day"], df["Crash"])
print(table)
```

```python
"""
H0: Day of week has NO effect on crash
H1: Day of week DOES affect crash
"""
```

```python
chi2, p_value, dof, expected = stats.chi2_contingency(table)

print(f"\nChi2 statistic : {chi2:.5f}")
print(f"P-value        : {p_value:.5f}")

if p_value < 0.05:
    print("Reject H0 ")
else:
    print("Fail to reject H0")
```
