```python
import pandas as pd
import yfinance as yf
from scipy import stats
```

## ANOVA

Compare 3 groups

```python
sp500 = yf.download("^GSPC", period="10y")[["Close"]]
usdx = yf.download("DX-Y.NYB", period="10y")[["Close"]]
eurusd = yf.download("EURUSD=X", period="10y")[["Close"]]

sp500.columns = ["SP500"]
usdx.columns = ["USDX"]
eurusd.columns = ["EURUSD"]


sp500_returns = sp500["SP500"].pct_change(fill_method=None)
usdx_returns = usdx["USDX"].pct_change(fill_method=None)
eurusd_returns = eurusd["EURUSD"].pct_change(fill_method=None)
```

```python
f_stat, p_value = stats.f_oneway(
    sp500_returns.dropna(),
    usdx_returns.dropna(),
    eurusd_returns.dropna(),
)

SIGNIFICANT_LEVEL = 0.05

print(f"F-Statistic : {f_stat:.5f}")
print(f"P-Value     : {p_value:.5f}")

if p_value < SIGNIFICANT_LEVEL:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

```python
print(f"SP500 mean   : {sp500_returns.mean()*100:.4f}%")
print(f"USDX mean    : {usdx_returns.mean()*100:.4f}%")
print(f"EURUSD mean  : {eurusd_returns.mean()*100:.4f}%")
```
