```python
import pandas as pd
import yfinance as yf
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

## Time Series

Time Series - data recorded over time in sequence

```python
eurcad = yf.download("EURCAD=X", period = "10y")[["Close"]]
eurcad.columns = ["EURCAD"]

eurcad["MA20"] = eurcad["EURCAD"].rolling(window=20).mean()
eurcad["MA50"] = eurcad["EURCAD"].rolling(window=50).mean()

eurcad[["MA20", "MA50"]].plot(figsize=(12,6))
plt.title("EURCAD with 20/50 MA")
plt.show()
```
