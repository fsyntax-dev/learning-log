```python
import pandas as pd
import yfinance as yf
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.graphics.tsaplots import plot_acf
```

## Autocorrelation

Correlation     - relationship between 2 different variable
Autocorrelation - relationship of 1 variabes with itself

```python
eurcad = yf.download("EURCAD=X", period="10y")[["Close"]]
eurcad.columns = ["EURCAD"]
eurcad_returns = eurcad["EURCAD"].pct_change(fill_method=None).dropna()

plot_acf(eurcad_returns, lags=30)
plt.title("EURCAD Daily Returns Autocorrelation")
plt.show()

```

```python
# squared return for volatility clustering
#   clustering focus on magnitude which ignore the direction
plot_acf(eurcad_returns**2, lags=30) 
plt.title("EURCAD Squared Daily Returns Autocorrelation")
plt.show()
```
