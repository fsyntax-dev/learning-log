```python
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
import seaborn as sns
```

# Distribution

Distribution        - how the data spread from the mean
Standard Deviation  - measurement of the spread
Skewness            - Shape of distribution - can be normal or skewed (left or right)
Kurtosis            - Shape of the mean - can be peaked or flat (Lepto, Meso, Platy)

```python
sp500 = yf.download("^GSPC", period="10y")[["Close"]]
sp500.columns = ["SP500"]
sp500_returns = sp500["SP500"].pct_change(fill_method=None)

sp500_returns.head()
```

```python
sp500_returns.plot(kind="hist", bins=50, figsize=(15,5))

plt.title("SP500 Daily Returns Distribution")
plt.xlabel("Daily Return")
plt.ylabel("Frequency")

plt.show()

print(f"Mean     : {sp500_returns.mean():.5f}")
print(f"Std Dev  : {sp500_returns.std():.5f}")
print(f"Skewness : {sp500_returns.skew():.5f}")
print(f"Kurtosis : {sp500_returns.kurt():.5f}")
```

## Skewness

Normal Distribution                     - Data spread equally left and right of mean
                                        - Outlier balanced equally

Skewed Distribution                     - Domination of outlier value pulled on a side 
Left Skewed (Negative Distribution)     - Tail left (Extreme Low Outlier)
Right Skewed (Positive Distribution)    - Tail right (Extreme High Outlier)

0 = Normal
- = Left Skewed
+ = Right Skewed

## Kurtosis

Kurtosis    - Shape of the mean
            - explain how often outlier happen

Mesokurtic  - Normal = 3

Leptokurtic - Peaked = Above 3 
Playtkurtic - Flat   = Below 3

## Kernel Density Estimate

KDE - smoothed version of distribution chart

Histogram depends on bins - change bins = change the shape
KDE does't care about bins - it just the true shape of data

```python
sns.histplot(sp500_returns.dropna(), 
             bins=50, 
             kde=True,
             color="steelblue",)

plt.title("SP500 Daily Returns Distribution")
plt.xlabel("Daily Return")
plt.ylabel("Frequency")
plt.show()
```
