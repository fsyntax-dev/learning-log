```python
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
```

## Simple Probability

Probability - how often does an event happen
P(A) = A / B
P(event) = number of the event happened / total observation

```python
sp500 = yf.download("^GSPC", period="10y")[["Close"]]
sp500.columns = ["SP500"]
sp500_returns = sp500["SP500"].pct_change(fill_method=None)

total_days = len(sp500.dropna())
crash_days = len(sp500_returns[sp500_returns < -0.03])

simple_crash_probability = crash_days/total_days

print(f"Total Days              : {total_days}")
print(f"Crash Days              : {crash_days}")
print(f"Probability of Crash    : {simple_crash_probability:.5f}")
print(f"Probability of Crash %  : {simple_crash_probability*100:.2f}%")


```

## Conditional Probability

Conditional Probability - updated probability based on new information

P(A | B) = P(A and B) / P(B)


```python
df = sp500_returns.to_frame()
df.columns = ["Today"]
df["Yesterday"] = df["Today"].shift(1)

yesterday_crash = df[df["Yesterday"] < -0.03]

both_crash = len(yesterday_crash[yesterday_crash["Today"] < -0.03])

conditional_crash_probability = both_crash / len(yesterday_crash)


print(f"Conditional Probability : {conditional_crash_probability*100:.2f} %")
```
