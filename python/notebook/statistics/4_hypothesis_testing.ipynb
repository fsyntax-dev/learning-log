```python
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
from scipy import stats
```

## Hypothesis Testing

Hypothesis Testing - it just answering the result either real or luck

H0 - nothing happen
H1 - something happen

p-value - probability value
significant level - acceptable probability (5% is industry norm)

to decide to pic H0 or H1 based on p-value

p < 0.05 -> reject H0
p > 0.05 -> fail to reject H0

To decide H0 or H1 we test with to get p-value
    T-test              - compare means
    2 sample T-test     - compare 2 group
    Anova               - compare 3 group
    Chi-square          - categorical variable

```python
sp500 = yf.download("^GSPC", period="10y")[["Close"]]
sp500.columns = ["SP500"]
sp500_returns = sp500["SP500"].pct_change(fill_method=None)

print(f"Daily Return: {sp500_returns.mean()*100:.2f}%")
print(f"Is {sp500_returns.mean()*100:.2f}% daily return real or fake?")
```

## T-test 

```python
"""
H0 : 0.05% Daily Return is Luck
H1 : 0.05% Daily Return is Real
"""

SIGNIFICANCE_LEVEL = 0.05

t_stat, p_value = stats.ttest_1samp(sp500_returns.dropna(),0)
print(f"T-statistic : {t_stat:.5f}")
print(f"P-value     : {p_value:.5f}")

if p_value < SIGNIFICANCE_LEVEL:
    print("Reject H0")
else:
    print("Fail to reject H0 ")
```
