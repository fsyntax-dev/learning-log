```python
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller
import seaborn as sns
```

```python
sp500 = yf.download("^GSPC", period="10y")[["Close"]]
sp500.columns = ["SP500"]
sp500_returns = sp500["SP500"].pct_change(fill_method=None)

usdx = yf.download("DX-Y.NYB", period="10y")[["Close"]]
usdx.columns = ["USDX"]
usdx_returns = usdx["USDX"].pct_change(fill_method=None)

# raw price - drifting result
adf_sp500 = adfuller(sp500.dropna())
print("Raw Price")
print(f"ADF Statisitc: {adf_sp500[0].round(5)}")
print(f"p value : {adf_sp500[1].round(5)}")

print("\n")

# daily return - stationary result
adf_sp500 = adfuller(sp500_returns.dropna())
print("Daily Return")
print(f"ADF Statisitc: {adf_sp500[0].round(5)}")
print(f"p value : {adf_sp500[1].round(5)}")

```

```python
df_returns = pd.concat([sp500_returns, usdx_returns], axis=1)
df_returns.columns = ["SP500_returns", "USDX_returns"]


sns.heatmap(df_returns.corr(), 
            annot=True, 
            cmap='coolwarm',
            vmin=-1, 
            vmax=1)
plt.title('Correlation Matrix')
plt.show()
```
