```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
```

## Bayes Statistics

Bayes - update your beliefs when new evidence arrives

Posterior = Prior Tau x Prior + Likelihood Tai x Likelihood / Prior Tau + Likelihood Tau

Prior       - belief before new evidence
Likelihood  - new evidence
Posterior   - updated belief after new evidence

Tau (τ)     - how much you trust your prior

### Prior

```python
sp500 = yf.download("SPY", period="5y")["Close"]["SPY"]
sp500_returns = sp500.pct_change(fill_method=None).dropna()

prior = sp500_returns.mean()
prior_tau = len(sp500_returns)

print(f"Prior Mean  : {prior:.5f} %")
print(f"Tau         : {prior_tau} days")
```

### Likelihood

```python
recent_returns = sp500_returns.tail(90)
likelihood = recent_returns.mean()
likelihood_tau = len(recent_returns)


print(f"Likelihood Mean : {likelihood:.5f} %")
print(f"Tau             : {likelihood_tau} days")
```

### Posterior

```python
posterior = (prior_tau * prior + likelihood_tau * likelihood) / (prior_tau + likelihood_tau)

print(f"Prior           : {prior:.5f}%")
print(f"Likelihood      : {likelihood:.5f}%")
print(f"Posterior       : {posterior:.5f}%")
```
