```python
import pandas as pd
import lseg.data as ld
import matplotlib.pyplot as plt
```

```python
ld.open_session()

df = ld.get_history("AAPL.O", interval="daily", fields="TRDPRC_1", start="2023-01-01", end="2025-12-31")
print(df)
```

```python
df["TRDPRC_1"].plot(figsize=(12,7), title= "AAPL 1H Close Price")
plt.xlabel("Period")
plt.ylabel("Price")

plt.tight_layout()
plt.show()
```
