```python
import datetime
import pandas as pd
import lseg.data as ld
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

```python
ld.open_session()

start_date = datetime.datetime.now() - datetime.timedelta(14)
end_date = datetime.datetime.now()
```

```python
df = ld.get_history(
    universe=["EUR=", "GBP=", "AUD="],
    fields=["MID_PRICE"],
    interval="30min",
    start=start_date,
    end=end_date,
)

print(df)
```

```python
plt.figure(figsize=(12,6))

df["AUD="].plot()

ax = plt.gca()


ax.set_xlabel("Date", loc="center")
ax.set_ylabel("Price", loc="center")
ax.set_title("AUD=")

plt.xticks(rotation=0, ha="center")


plt.show()
```
